````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("/Users/almatiliusyz/Almat/Lessons/DSA/dsa-homework/orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````

````python
# Маркетинг предлагает скидку 15% на оплаченные заказы дороже 50 000 ₸. Сколько выручки магазин недополучит?
# Посчитайте дважды: со всеми заказами и без заказов клиента TEST. Почему ответы различаются в несколько раз?
paid_orders = paid
amount_more_50 = ( amount > 50000)

total_orders = paid_orders & amount_more_50
total_orders
````

````python
with_sale = amount[total_orders].sum() * 0.15
print(with_sale)
````

````python
no_test = cust != "TEST"
total_orders_no_test = total_orders & no_test

with_sale_no_test = amount[total_orders_no_test].sum() * 0.15
print(with_sale_no_test)
````

````python
# 5. Для **каждого** города — доля заказов, доставленных за 1–2 дня. Список городов: `np.unique(city)`; цикл по 6 городам здесь можно.
cities = np.unique(city)

for c in cities:
    city_mask = city == c
    delivered_fast = days <= 2

    total_in_city = city_mask.sum()
    fast_in_city = (city_mask & delivered_fast).sum()

    share = fast_in_city / total_in_city
    print(c, share)
````

````python
avg = amount[paid & (cust != "TEST")].mean()
print("Средний чек:", round(avg))

# Возможно надо брать оплаченные заказы и не тестовые
````
