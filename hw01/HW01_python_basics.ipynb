### База
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.


````python
cities_clean = []

for city in cities:
    clean_city = city.strip().title()
    cities_clean.append(clean_city)

print(cities_clean)
````

````python
uniq_city = list(set(cities_clean))
print(uniq_city)
    
````

2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.

````python
amounts_int = []

for amount in amounts:
    amounts_int.append(int(amount))

print(amounts_int)
````

3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).

````python
total_orders = len(amounts_int)
total_sum = sum(amounts_int)
avg_check = round(total_sum / total_orders)

print("Общее число заказов: ", total_orders)
print("Общая сумма заказов: ", total_sum)
print("Средний чек: ", avg_check)
````

4. Сколько заказов не оплачено и на какую сумму?

````python
unpaid_count = 0
unpaid_sum = 0
for c, a, p in zip(cities, amounts, paid):
    if p == False:
        unpaid_count += 1
        unpaid_sum += int(a)
print("Количество не оплаченых заказов", unpaid_count)
print("Сумма не оплаченого заказа:", unpaid_sum)
````

6. Выручка (только оплаченные) по Алматы и по Астане отдельно.

````python
almaty_sum = 0
astana_sum = 0
for c, a, p in zip(cities_clean, amounts_int, paid):
    if p and c == 'Almaty':
        almaty_sum += (a)
    elif p and c == 'Astana':
        astana_sum += (a)

print("Выручка (только оплаченные) по Алматы", almaty_sum)
print("Выручка (только оплаченные) по Астане:", astana_sum)
````

7. Номер (индекс) и город самого крупного **оплаченного** заказа.

````python
max_amount = 0
max_index = 0

for i in range(len(amounts_int)):
    if paid[i] and amounts_int[i] > max_amount:
        max_amount = amounts_int[i]
        max_index = i

print("Наивысшая сумма: ", max_amount)
print("Наивысшый индекс: ",max_index)
````

8. Доля заказов выходного дня (`сб`, `вс`) в процентах.


````python
weekend_count = 0
for day in days:
    if day == "сб" or day == "вс":
        weekend_count += 1

weekend_share = weekend_count / len(days) * 100
print("8. Доля заказов выходного дня в процентах: ", weekend_share)
````

9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
saved = 0
months = 0
while saved < 400000:
    saved += 15000
    months += 1
print("Количество месяцов: ", months)
print("Количество лет: ",months / 12)
````

10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.

````python
unique_cities = []
revenue_by_city = []

for city in cities_clean:
    if city not in unique_cities:
        unique_cities.append(city)
        revenue_by_city.append(0)

for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid:
        locator = unique_cities.index(city)
        revenue_by_city[locator] += amount

print(unique_cities)
print(revenue_by_city)
````

````python
n = len(revenue_by_city)
for i in range(n):
    for j in range(n - 1 - i):
        if revenue_by_city[j] < revenue_by_city[j + 1]:
            revenue_by_city[j], revenue_by_city[j + 1] = revenue_by_city[j + 1], revenue_by_city[j]
            unique_cities[j], unique_cities[j + 1] = unique_cities[j + 1], unique_cities[j]

print(unique_cities)
print(revenue_by_city)
````

11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.

````python
sorted_amounts = sorted(amounts_int)
n = len(sorted_amounts)
if n % 2 == 1:
    median = sorted_amounts[n // 2]
else:
    mid1 = sorted_amounts[n // 2 - 1]
    mid2 = sorted_amounts[n // 2]
    median = (mid1 + mid2) / 2
print(median)
````

12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```

````python
unique_days = []
sum_by_day = []
for day in days:
    if day not in unique_days:
        unique_days.append(day)
        sum_by_day.append(0)
for day, amount, is_paid in zip(days, amounts_int, paid):
    if is_paid:
        locator = unique_days.index(day)
        sum_by_day[locator] += amount

for day, summ in zip(unique_days, sum_by_day):
    bars = "█" * (summ // 5000)
    print(f"{day} | {bars} {summ}")
````
