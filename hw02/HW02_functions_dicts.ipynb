1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).

````python
def to_num(s):
    return float(s.replace(" ", ""))
````

````python
def clean_city(s):
    s = s.strip().title()
    if s == "Almaty":
        return "Алматы"
    if s == "Astana":
        return "Астана"
    if s == "Shymkent":
        return "Шымкент"
    return s

    
cities_after = set()
for r in rows:
    cities_after.add(clean_city(r["city"]))
print(cities_after)

````

2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.

````python
category_count = {}
for r in rows:
    cat = r["category"]
    category_count[cat] = category_count.get(cat, 0) + 1

print(category_count)
````

3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.

````python
category_revenue = {}
category_paid_count = {}
for r in rows:
    if r["status"] == "оплачен":
        cat = r["category"]
        category_revenue[cat] = category_revenue.get(cat, 0) + to_num(r["amount"])
        category_paid_count[cat] = category_paid_count.get(cat, 0) + 1

print(category_revenue)
print(category_paid_count)
````

4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
month_count = {}
for r in rows:
    month = r["order_date"][:7]
    month_count[month] = month_count.get(month, 0) + 1

print(month_count)
````

5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.

````python
month_revenue = {}
for r in rows:
    if r["status"] == "оплачен":
        month = r["order_date"][:7]
        month_revenue[month] = month_revenue.get(month, 0) + to_num(r["amount"])
change_pct = (month_revenue["2026-10"] - month_revenue["2026-09"]) / month_revenue["2026-09"] * 100

print("Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах:", change_pct)
````

6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.

````python
city_total = {}
city_cancelled = {}
for r in rows:
    city = clean_city(r["city"])
    city_total[city] = city_total.get(city, 0) + 1
    if r["status"] == "отменён":
        city_cancelled[city] = city_cancelled.get(city, 0) + 1

for city in city_total:
    cancelled = city_cancelled.get(city, 0)
    total = city_total[city]
    rate = cancelled / total * 100
    print(f"{city}: {share:.1f}%")

print(city_total)
print(city_cancelled)
````

7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.

````python
customer_revenue = {}
for r in rows:
    if r["status"] == "оплачен":
        cid = r["customer_id"]
        customer_revenue[cid] = customer_revenue.get(cid, 0) + to_num(r["amount"])
top5 = sorted(customer_revenue.items(), key=lambda t: t[1], reverse=True)[:5]

print(top5)
````

9. Найдите полные дубли строк: превратите каждую строку в `tuple(r.values())` и проверяйте через множество. Сколько их и в каком месяце?

````python
seen = set()
duplicates = []
for r in rows:
    key = tuple(r.values())
    if key in seen:
        duplicates.append(r)
    else:
        seen.add(key)

print("Дублей найдено:", len(duplicates))

dup_months = {}
for r in duplicates:
    month = r["order_date"][:7]
    dup_months[month] = dup_months.get(month, 0) + 1

print("По месяцам:", dup_months)
````

10. Пересчитайте выручку по месяцам без дублей и без клиента из задачи 7. Как изменился процент из задачи 5?

````python
top_customer_id = top5[0][0]
seen2 = set()
month_revenue_clean = {}
for r in rows:
    key = tuple(r.values())
    if key in seen2:
        continue
    seen2.add(key)
    if r["status"] == "оплачен" and r["customer_id"] != top_customer_id:
        month = r["order_date"][:7]
        month_revenue_clean[month] = month_revenue_clean.get(month, 0) + to_num(r["amount"])
change_pct2 = (month_revenue_clean["2026-10"] - month_revenue_clean["2026-09"]) / month_revenue_clean["2026-09"] * 100

print(change_pct2)
````

11. У какого клиента больше всего заказов (после удаления дублей)? Сколько?

````python
seen3 = set()
customer_order_count = {}
for r in rows:
    key = tuple(r.values())
    if key in seen3:
        continue
    seen3.add(key)
    cid = r["customer_id"]
    customer_order_count[cid] = customer_order_count.get(cid, 0) + 1
top_by_count = max(customer_order_count, key=customer_order_count.get)

print(top_by_count)
````
