````python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
````

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База
# 1. Напишите функции `to_num(s)` и `clean_city(s)` 
# (6 городов, латиница и разный регистр → кириллица с заглавной).

def to_num(s):
    return int(float(s.strip().replace(" ", "")))

def clean_city(s):
    s = s.strip().title()
    if s == "Almaty":
        return "Алматы"
    if s == "Astana":
        return "Астана"
    if s == "Shymkent":
        return "Шымкент"
    return s    
````

````python
# 2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.

category_count = {}

for r in rows:
    category = r["category"]
    category_count[category] = category_count.get(category, 0) + 1

for c, v in sorted(category_count.items(), key=lambda t: t[1], reverse=True):
    print(f"{c:<11} {v:>7}")
````

````python
# 3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.

category_rev = {}
category_count = {}

for r in rows:
    category = r["category"]
    amount = to_num(r["amount"])
    paid = r["status"]

    if paid == 'оплачен':    
        category_rev[category] = category_rev.get(category, 0) + amount
        category_count[category] = category_count.get(category, 0) + 1

print(f"{'Категория':<15} {'Выручка':>12} {'Количество':>8} {'Средний чек':>15}")

for category in category_rev:
    revenue = category_rev[category]
    count = category_count[category]
    average = revenue / count

    print(f"{category:<15} {revenue:>12,.0f} {count:>8} {average:>15,.2f}")
````

````python
# 4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

month_count = {}
for r in rows:
    if r["status"] == "оплачен":
        month = r["order_date"][:7]
        month_count[month] = month_count.get(month, 0) + 1
        
month_count
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
# 5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
month_rev = {}
for r in rows:
    amount = to_num(r["amount"])
    if r["status"] == "оплачен":
        month = r["order_date"][:7]
        month_rev[month] = month_rev.get(month, 0) + amount
        
diff_per = (month_rev["2026-10"] - month_rev["2026-09"]) / month_rev["2026-09"] * 100
print(f"Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах {diff_per}")
````

````python
# 6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.

city_total = {}
city_cancelled = {}
for r in rows:
    city = clean_city(r["city"])
    city_total[city] = city_total.get(city, 0) + 1
    if r["status"] == "отменён":
        city_cancelled[city] = city_cancelled.get(city, 0) + 1

for city in city_total:
    cancelled = city_cancelled.get(city, 0)
    total = city_total[city]
    rate = cancelled / total * 100
    print(f"{city}: {rate:.1f}%")

print(city_total)
print(city_cancelled) 
````

````python
# 7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. 
# Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.

customers_rev = {}
for  r in rows:
    customers = r["customer_id"]
    if r["status"] == "оплачен":
        customers_rev[customers] = customers_rev.get(customers, 0) + 1
# print(customers_rev)

print("Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов")
for c, v in sorted(customers_rev.items(), key=lambda t: t[1], reverse=True)[:5]:
    print(f"{c:<10} {v:>7,}")


best = max(customers_rev, key=customers_rev.get)
print("Клиент с наивысшим количеством заказа:", best, customers_rev[best])
````
